## 1. Why Part 2 Exists in This Course

In Part 1, you built powerful perception capabilities:

- A YOLO model detects objects in a frame.
- A pose estimator identifies body keypoints.
- A tracker maintains identities across video frames.

These are *tools*—but tools alone don’t make decisions. A retail analytics system doesn’t just detect people; it needs to decide *when* to send an alert, *how* to summarize the day’s traffic, and *whether* to escalate anomalies. That decision layer is where **agents** come in.

An agent is a system that:
1. **Observes** its environment (e.g., receives an image, a sensor reading, or a user message).
2. **Decides** what to do next (call a tool? ask a clarifying question? produce a report?).
3. **Acts** on that decision.
4. **Updates** its state and repeats.

Part 2 teaches you to think at this systems level—not just "how do I run a model?" but "how do I build a loop that uses models intelligently?"

## 3. How This Connects to Part 1 (Computer Vision)

Part 1 gave you *perception tools*. Part 2 shows you how to *orchestrate* them. Here’s a concrete example of how the pieces fit together:

**Scenario: Automated Store Entrance Monitoring**

1. **Perception (Part 1)**: A YOLO tracker counts people entering and exiting a store, frame by frame.
2. **Summarization (Part 2)**: A prompt instructs an LLM to summarize hourly counts into a human-readable report.
3. **Decision (Part 2)**: If the count exceeds a threshold, the agent decides to send an alert.
4. **Action (Part 2)**: The agent formats and dispatches the alert (in a real system, this could be email, Slack, or a dashboard update).
5. **Memory (Part 2)**: The agent remembers today’s totals to compare with yesterday.
6. **Evaluation (Part 2)**: You define test cases: “If count > 500, alert should fire.” You run regression tests.

Your CV models become *tools* that an agent can invoke. The agent loop ties everything together into a coherent, goal-directed system.

> **Key insight**: The CV model doesn’t change. What changes is *how you use it*—wrapping it in a decision-making layer.
